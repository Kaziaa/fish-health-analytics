# summary
The 2025 BarentsWatch lice dataset contained 55,718 observations from 1,082 localities. Of these, 30,297 observations had both a reported adult female lice value and a valid weekly lice limit. The overall breach rate was 3.52%. For locality-level ranking, I required at least 26 valid weekly observations, which retained 660 localities. # were the breaches isolated, or did they occur in sustained episodes?

# measure the gaps between breach weeks

```python
breach_series = pd.Series(breach_weeks)

breach_series.diff()
```

```python
# mark where a new episode starts
new_episode = breach_series.diff().ne(1)

new_episode
```

```python
episode_id = new_episode.cumsum()

episode_id
```

```python
episodes = pd.DataFrame({
    "week": breach_series,
    "episode": episode_id
})

episodes
```

```python
episode_summary = (
    episodes
    .groupby("episode")["week"]
    .agg(
        start_week="min",
        end_week="max",
        breach_weeks="count"
    )
)

episode_summary
```

```python
episode_summary["breach_weeks"].sum()
```

# Summary
Urda had 27 breach weeks across 43 valid lice observations in 2025, corresponding to a breach rate of 62.79%. These breaches occurred in six distinct episodes. The longest episode lasted eight consecutive weeks, from week 4 to week 11. Two additional prolonged breach periods occurred during weeks 34–40 and 42–48, each lasting seven weeks. The analysis also showed that breach status depends on the applicable weekly regulatory limit, which temporarily dropped from 0.5 to 0.2 during weeks 16–21.
